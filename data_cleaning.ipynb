## Fix team names

```python
team_fix = {
    'Delhi Daredevils': 'Delhi Capitals',
    'Kings XI Punjab': 'Punjab Kings',
    'Royal Challengers Bangalore': 'Royal Challengers Bengaluru',
    'Rising Pune Supergiants': 'Rising Pune Supergiant',
    'Deccan Chargers': 'Sunrisers Hyderabad'
}

for col in ['team1', 'team2', 'toss_winner', 'winner']:
    df[col] = df[col].replace(team_fix)
```

```python
df.sample(10)
```

```python
df['team1'].unique().tolist()
```

## Fix venue names

```python
df['venue'] = df['venue'].str.split(',').str[0].str.strip()
```

```python
venue_fix = {
    'M.Chinnaswamy Stadium': 'M Chinnaswamy Stadium',
    'Punjab Cricket Association Stadium': 'Punjab Cricket Association IS Bindra Stadium',
    'Feroz Shah Kotla': 'Arun Jaitley Stadium',          # renamed 2019
    'Sardar Patel Stadium': 'Narendra Modi Stadium',      # renamed 2021
    'Subrata Roy Sahara Stadium': 'Maharashtra Cricket Association Stadium',
    'Zayed Cricket Stadium': 'Sheikh Zayed Stadium',
    'Brabourne Stadium': 'Brabourne Stadium'
}

df['venue'] = df['venue'].replace(venue_fix)
```

```python
df.sample(10)
```

```python
df['season'] = df['season'].str.split('/').str[0].astype(int)
```

```python
df.sample(10)
```
